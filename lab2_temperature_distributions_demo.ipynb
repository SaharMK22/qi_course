## Step 1 -- Predict

Before looking at any data: think about July daily high temperatures in
**Tucson**, **Buffalo**, and **Seattle**.

- Which city do you expect to have the *most* day-to-day variability in

  above or below that city's typical July high? Is that rare or

(In the real lab, students would write a short prediction here before
running any code -- the point is committing to an expectation first, so

```
# Step 2 -- Calculate: load the data and compute basic statistics.
#
# NOTE ON THE DATA: load_demo_july_highs() currently returns SYNTHETIC
# data -- daily highs drawn directly from a normal distribution with a
# realistic mean/std for each city. It is a stand-in so the rest of the
# pipeline (stats, plotting, interpretation) can be built and tested now.
# Before this lab goes live, this should be replaced with a real,
# multi-decade station record (e.g., NOAA's GHCN-Daily archive), so the
# "is it approximately Gaussian?" question is something students
# discover in real observations, not something built into the demo.

city = "Tucson"  # try "Buffalo" or "Seattle" too
temps = qc.load_demo_july_highs(city=city, n_years=100)

stats = qc.descriptive_stats(temps["high_temp_c"])
for key, value in stats.items():
    print(f"{key}: {value:.2f}" if isinstance(value, float) else f"{key}: {value}")
```

**Check your prediction against the numbers.** Was the mean close to
what you expected? Is the skewness close to 0 (consistent with a
symmetric, Gaussian-like shape), or clearly off to one side?

```
# Step 3 -- Visualize: histogram of the data with a fitted normal
# curve overlaid, so the comparison is visual, not just numbers.

grid, pdf, params = qc.fit_normal_pdf(temps["high_temp_c"])

qc.histogram_with_fit(
    temps["high_temp_c"], grid, pdf,
    title=f"{city} July daily highs (synthetic) vs. fitted normal",
    xlabel="High temperature (C)",
)

print(f"Fitted normal: mean = {params['mean']:.1f} C, std = {params['std']:.2f} C")
```

## Step 4 -- Explain

- Does the histogram look close to the fitted normal curve, or are
  there clear differences (e.g., a longer tail on one side)?
- Using the fitted mean and standard deviation: about how often would
  you expect a July day in this city to be *more than 2 standard
  deviations* above the mean? (This is a good place to connect back to
  the empirical rule, ~95% within 2 std, without deriving it from
  scratch.)
- If this were a real, observed record instead of synthetic data, what
  would a `skewness` clearly different from 0 suggest about the
  physical process generating hot days in this location (e.g.,
  heat-wave events pushing a long tail on the hot side)?
- Try changing `city` to `"Buffalo"` or `"Seattle"` above and re-run.
  Does the *shape* of the fit change, or mainly the mean and spread?
