The next two cells just exercise one function from each of the two
labs we've prototyped so far, so you can see real output, not just an

```
# Cell 3 -- Lab 1 preview: a synthetic rainfall time series.

# load_demo_harvey_precip() hands back a table (a pandas DataFrame)
# with two columns: "hour" (0 through 119 -- five days of hourly
# readings) and "rain_mm_per_hr" (how hard it's raining at that hour).
# It's synthetic data shaped like a slow-moving storm -- rain builds,
# stays heavy for a while, then tapers off -- standing in for a real
# rain-gauge record until one is wired in.
df = qc.load_demo_harvey_precip()

# quick_timeseries_plot just draws rain_mm_per_hr against hour with
# sensible defaults (line width, gridlines, axis labels) so every
# notebook's plots look the same without rewriting that formatting
# code each time. It hands back the plot's Axes object in case you
# want to keep customizing the figure afterward.
qc.quick_timeseries_plot(
    df["hour"], df["rain_mm_per_hr"],
    title="Synthetic rainfall pulse (Lab 1 placeholder data)",
    xlabel="Hour", ylabel="Rain (mm/hr)",
)
```

```
# Cell 4 -- Lab 2 preview: synthetic July daily highs for Tucson,
# with a fitted normal curve overlaid.

# load_demo_july_highs() returns 100 years of simulated July daily
# high temperatures for one city (here, Tucson) -- one row per day.
# It's synthetic, drawn directly from a normal distribution by
# construction, standing in until a real long-term station record
# replaces it.
temps = qc.load_demo_july_highs(city="Tucson", n_years=100)

# fit_normal_pdf takes that column of temperatures and fits a normal
# (Gaussian) curve to it, the same way you'd estimate a mean and a
# standard deviation by hand. It returns three things: a grid of
# x-values to draw the curve over, the curve's height at each of those
# x-values, and the fitted mean/std themselves in a small dictionary,
# so you can see exactly what was estimated from the data.
grid, pdf, params = qc.fit_normal_pdf(temps["high_temp_c"])

# histogram_with_fit draws the actual data as a histogram and lays the
# fitted curve on top of it, so you can see with your own eyes how
# well a normal distribution describes this particular sample.
qc.histogram_with_fit(
    temps["high_temp_c"], grid, pdf,
    title="Synthetic Tucson July daily highs vs. fitted normal",
    xlabel="High temperature (C)",
)
print(f"Fitted mean: {params['mean']:.1f} C, fitted std: {params['std']:.2f} C")
```

## If something breaks

Cell 1 (the install) is the one most likely to act up, and it's almost
always one of two things: either the `v0.2.0` tag doesn't match what's
on GitHub, or Colab's runtime needs a fresh restart (Runtime > Restart
session) after the install finishes. If Cell 2 runs but the version it
prints isn't `0.2.0`, a restart usually fixes it.

Beyond that, don't spend time guessing what went wrong. Copy the exact
error text, note which cell it happened in, and send it to me -- that's
almost always enough for me to track it down.
